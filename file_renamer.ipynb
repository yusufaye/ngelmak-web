```python
import re
import os
from glob import glob
from typing import List
```

```python
path = 'src/app/entities/ngelmak-account-privilege/'
replacewith = [ 'ngelmak-account-privilege', 'user-privilege', ]
replaceOnContentWith = [
  [ 'INgelmakAccountPrivilege', 'IUserPrivilege' ],
  [ 'NgelmakAccountPrivilege', 'UserPrivilege' ],
  [ 'ngelmakAccountPrivilege', 'userPrivilege' ]
]

def get_paths(directory: str, pattern: str) -> List[str]:
  paths = [ os.path.join(root, file) for root, _, files in os.walk(directory) for file in files ]
  return [ path for path in paths if re.search(r'%s'%(pattern), path) ]

# rename files
for filename in get_paths(directory=path, pattern=replacewith[0]):
  new_filename = filename.replace(*replacewith)
  os.rename(src=filename, dst=new_filename)
# replace on content
for file in get_paths(directory=path, pattern=replacewith[1]):
  lines = []
  with open(file) as f:
    for line in f.readlines():
      for old, new in replaceOnContentWith:
        line = line.replace(old, new)
      lines.append(line)
  with open(file, mode='w') as f:
    f.writelines(lines)
```
